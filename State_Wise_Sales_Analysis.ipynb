# State/Product/Gender Wise Sales Analysis:

```python
import matplotlib.pyplot as plt
customer_names = ["Ankit","Rohan","Mannat","Rajeev","Shakshi","Preetam",
                 "Neha","Kartik","Radhika","Hemant"]
products = ["Mobile","Laptop","Ac","Mobile","Computer","Calculator","Ac",
            "Computer","Mobile","Calculator"]
genders = ["Male","Male","Female","Male","Female","Male","Female","Male",
           "Female","Male"]
states = ["Hp","Mp","Up","Punjab","Mp","Delhi","Hp","Up","Punjab","Delhi"]
prices = [40000,50000,60000,30000,50000,10000,80000,40000,20000,25000]
```

```python
unique_states = []
state_wise_sales = []
for state in states:
  if state not in unique_states:
     unique_states.append(state)
     state_wise_sales.append(0)
print(unique_states)
print(state_wise_sales)

for i in range(len(states)):
    index = unique_states.index(states[i])
    state_wise_sales[index] = state_wise_sales[index] + prices[i]
print("state wise sales:", state_wise_sales)

plt.title("State Wise Sales")
plt.xlabel("States")
plt.ylabel("Sales")
plt.bar(unique_states,state_wise_sales,color = ["red","green","blue"])
plt.show()
```

```python
unique_products = []
product_wise_sales = []
for product in products:
  if product not in unique_products:
     unique_products.append(product)
     product_wise_sales.append(0)
print(unique_products)
print(product_wise_sales)
for i in range(len(products)):
    index = unique_products.index(products[i])
    product_wise_sales[index] = product_wise_sales[index] + prices[i]
print("product wise sales:", product_wise_sales)

plt.title("Product Wise Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.bar(unique_products,product_wise_sales,color = ["green","purple","red"])
plt.show()
```

```python
unique_genders = []
gender_wise_sales = []
for gender in genders:
  if gender not in unique_genders:
     unique_genders.append(gender)
     gender_wise_sales.append(0)

for i in range(len(genders)):
    index = unique_genders.index(genders[i])
    gender_wise_sales[index] = gender_wise_sales[index] + prices[i]
print("Gender Wise Sales:", gender_wise_sales)

plt.title("Gender Wise Sales")
plt.xlabel("Genders")
plt.ylabel("Sales")
plt.bar(unique_genders,gender_wise_sales,color = ["red","green","blue"])
plt.show()
```

# State/Products/Gender Wise Sales Analysis On a Dataset:

```python
import matplotlib.pyplot as plt
import csv
name,gender,state,product,amount=[],[],[],[],[]
with open("sales_data.csv","r")as file:
   reader=csv.reader(file)
   header=next(reader)
   for row in reader:
     name.append(row[1])
     gender.append(row[2])
     state.append(row[3])
     product.append(row[5])
     amount.append(int(row[7]))
```

```python
unique_state=[]
state_wise_sale=[]
for i in state:
   if i  not in unique_state:
     unique_state.append(i)
     state_wise_sale.append(0)
for i in range(len(state)):
  index1=unique_state.index(state[i])
  state_wise_sale[index1] = state_wise_sale[index1]+amount[i]
print(unique_state)
print(state_wise_sale)
plt.figure(figsize = (8,8))
plt.title("State Wise Sale")
plt.xlabel("States")
plt.ylabel("Sales")
plt.bar(unique_state,state_wise_sale,color = ["red","green","blue"])
plt.show()
```

```python
plt.figure(figsize = (8,8))
unique_products=[]
product_wise_sale=[]
for i in product:
   if i not in unique_products:
     unique_products.append(i)
     product_wise_sale.append(0)
for i in range(len(product)):
   index2=unique_products.index(product[i])
   product_wise_sale[index2]=product_wise_sale[index2]+amount[i]
print(unique_products)
print(product_wise_sale)
plt.figure(figsize = (9,9))
plt.title("Product Wise Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.bar(unique_products,product_wise_sale,color = ["purple","green","red"])
plt.show()
```

```python
unique_genders=[]
gender_wise_sale=[]
for i in gender:
   if i not in unique_genders:
     unique_genders.append(i)
     gender_wise_sale.append(0)
for i in range(len(gender)):
   index2=unique_genders.index(gender[i])
   gender_wise_sale[index2] = gender_wise_sale[index2]+amount[i]
print(unique_genders)
print(gender_wise_sale)
plt.title("Gender Wise Sale")
plt.xlabel("Gender")
plt.ylabel("Sales")
plt.bar(unique_genders,gender_wise_sale,color = ["purple","green"])
plt.show()
```

## State/Product/Gender Wise Sales Analysis By Using Pandas.
