```python
class Employee:
  def __init__(self,p,r,t): # constructor
    self.p=p
    self.r=r
    self.t=t
    si=p*r*t/100
    print(si)
  def emp_info(self):# self is used so that object can call to the class.
     print("name","raju")
  def salary(self):
     print("100000")
  def age(self):
     print("18")
object=Employee(1000,4,2)
object.emp_info()
object.age()
print(object.p)
```

```python
# multilabel inheritance:
class Item():
  def first_item_name(self):
    print("first item name","pizza")
  def second_item_name(self):
    print("second item name","burger")
  def third_item_name(self):
    print("third item name","tea")
    print("**********************")
  def item1_price(self):
    print("item1 price",300)
  def item2_price(self):
    print("item2 price",50)
  def item3_price(self):
    print("item3 price",30)
    print("**********************")
class Order(Item):
  def first_order(self):
    print("first order","pizza")
  def second_order(self):
    print("second order","burger")
  def third_order(self):
    print("third_order","tea")
    print("**********************")
class User(Order):
  def total_bill(self):
    total_bill=0
    item1_price=300
    item2_price=50
    item3_price=30
    total_bill=total_bill+item1_price+item2_price+item3_price
    print("total_bill",total_bill)
    discount=10
    less_bill=total_bill*discount/100
    print("less_bill",less_bill)
    payble_bill=total_bill-less_bill
    print("payble_bill",payble_bill)
object=User()
object.first_item_name()
object.second_item_name()
object.third_item_name()
object.item1_price()
object.item2_price()
object.item3_price()
object.first_order()
object.second_order()
object.third_order()
object.total_bill()
```

```python
# multiple inheritance:
class Items():
   def item1(self):
     print("item1","pizza")
   def item2(self):
     print("item2","burger")
   def item3(self):
     print("item3","tea")
     print("****************")
   def item1_price(self):
     print("item1 price",300)
   def item2_price(self):
     print("item2 price",50)
   def item3_price(self):
     print("item3 price",30)
     print("***************")
class Order():
   def order1(self):
     print("order1","pizza")
   def order2(self):
     print("order2","burger")
   def order3(self):
     print("order3","tea")
     print("*********")
class bill(Items,Order):
   def bill(self):
    total_bill=0
    item1_price=300
    item2_price=50
    item3_price=30
    total_bill=total_bill+item1_price+item2_price+item3_price
    print("total_bill",total_bill)
    discount=10
    less_bill=total_bill*discount/100
    print("less_bill",less_bill)
    payble_bill=total_bill-less_bill
    print("payble_bill",payble_bill)
    print("****************")
object=bill()
object.item1()
object.item2()
object.item3()
object.item1_price()
object.item2_price()
object.item3_price()
object.bill()
```

```python
# hybrid inheritance:
class Items():
   def item1(self):
     print("item1","pizza")
   def item2(self):
     print("item2","burger")
   def item3(self):
     print("item3","tea")
   def item1_price(self):
     print("item1",300)
   def item2_price(self):
     print("item2",50)
   def item3_price(self):
     print("item3",30)
class Order(Items):
   def order1(self):
     print("order1","pizza")
   def order2(self):
     print("order2","burger")
class Bill(Order,Items):
   def total_bill(self):
    total_bill=0
    item1_price=300
    item2_price=50
    item3_price=30
    total_bill=total_bill+item1_price+item2_price+item3_price
    print("total_bill",total_bill)
    discount=10
    less_bill=total_bill*discount/100
    print("less_bill",less_bill)
    payble_bill=total_bill-less_bill
    print("payble_bill",payble_bill)
class Receipt(Bill):
   def receipt(self):
     print("receipt","received")
object=Receipt()
object.receipt()
object.total_bill()
```

```python
from abc import ABC,abstractmethod

class Payment(ABC):
  @abstractmethod
  def payment(self,method,amount):
    pass
class UpiPayment(Payment):
      def payment(self,method,amount):
        print(f"{amount} done by {method}")
class CreditCardPayment(Payment):
      def payment(self,method,amount):
        print(f"{amount} done by {method}")
class DebitCardPayment(Payment):
      def payment(self,method,amount):
           print(f"{amount} done by {method}")
object1=UpiPayment()
object2=CreditCardPayment()
object3=DebitCardPayment()

def payment_process(method):
  if method=="Upi":
    return UpiPayment()
  elif method=="Credit":
    return CreditCardPayment()
  else:
    return DebitCardPayment()
pay_object=payment_process("Upi")
pay_object.payment("Gpay",10000)
```
